```python
from PIL import Image
import cv2
import numpy as np
from matplotlib import pyplot as plt
import skimage as ski
```

```python
img = Image.open('jiedao.jpg')
img_gray = img.convert('L')
img_gray
```

```python
plt.imshow(img_gray,cmap='gray')
plt.axis('off')
```

```python
img_array = np.array(img_gray)
plt.hist(img_array.ravel(), 256, [0, 256])
plt.title('Grayscale Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.show()
```

```python
# 设置阈值
threshold_value = 100  # 你可以根据需求更改此值

# 应用阈值分割
_, thresholded_image = cv2.threshold(img_array, threshold_value, 255, cv2.THRESH_BINARY)

# 显示原始图像和阈值分割后的图像
plt.figure(figsize=(10, 5))

# 原始图像
plt.subplot(1, 2, 1)
plt.imshow(img_array, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# 阈值分割后的图像
plt.subplot(1, 2, 2)
plt.imshow(thresholded_image, cmap='gray')
plt.title('Thresholded Image')
plt.axis('off')

plt.show()
```

```python
thresholdeds = [50,100,200]

multi_threshold_img = np.zeros_like(img_array)

multi_threshold_img[img_array<thresholdeds[0]] = 0
multi_threshold_img[(img_array>thresholdeds[0]) & (img_array<thresholdeds[1])] = 128
multi_threshold_img[img_array>thresholdeds[2]] = 255

# 原始图像
plt.subplot(1, 2, 1)
plt.imshow(img_array, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# 阈值分割后的图像
plt.subplot(1, 2, 2)
plt.imshow(multi_threshold_img, cmap='gray')
plt.title('Thresholded Image')
plt.axis('off')

plt.show()
```
