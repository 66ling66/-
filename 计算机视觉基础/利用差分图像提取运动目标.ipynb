```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.ones((10,10))
x = np.pad(x,pad_width=1,mode='constant',constant_values=1)
x = np.pad(x,pad_width=5,mode='constant',constant_values=0)


size = 10

# # 初始化一个全零矩阵
# x = np.zeros((size, size), dtype=int)

# # 构造右下角为直角的三角形
# for i in range(size):
#     for j in range(i + 1):  # 条件：列索引小于等于行索引
#         x[i, j] = 1

x = np.pad(x,pad_width=5,mode='constant',constant_values=0)
n = 1

# 创建一个与 x 相同大小的空数组，初始化为 0
result = np.zeros_like(x)

# 通过切片将 1 的像素往右和往上移动 n 格
result[n:, :-n] = x[:-n, n:]
```

```python
plt.subplot(1,2,1)
plt.imshow(x,cmap='gray')
plt.axis('off')
plt.title('orginal image')
plt.subplot(1,2,2)
plt.imshow(result,cmap='gray')
plt.axis('off')
plt.title('moved image')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 假设 x 是原始图像，result 是移动后的图像
# 计算差图像
difference = np.abs(x - result)

# 绘制原始图像、移动后图像和差图像
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(x, cmap='gray')
plt.axis('off')
plt.title('Original Image')

plt.subplot(1, 3, 2)
plt.imshow(result, cmap='gray')
plt.axis('off')
plt.title('Moved Image')

plt.subplot(1, 3, 3)
plt.imshow(difference, cmap='gray')
plt.axis('off')
plt.title('Difference Image')

plt.suptitle('Motion Detection Using Difference Image')
plt.show()
```
