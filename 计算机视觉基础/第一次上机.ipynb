```python
import cv2
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import copy
```

$$Gray=0.2989×Blue+0.5870×Green+0.1140×Red$$

```python
def image_to_gray_cv2(filename):
    image = cv2.imread(filename)
    gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
    cv2.imshow('gray_image',gray_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
def image_to_gray_pil(filename):
    image = Image.open(filename)
    image = image.convert("L")
    plt.imshow(image,cmap='gray')
    plt.show()
def binary_image(image,threshold):
    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            if image[i][j]>threshold:
                image[i][j] = 255
            else:
                image[i][j] = 0
    return image
```

```python
imagex = copy.deepcopy(np.array(image))
```

```python
image1 = binary_image(imagex,100)
image2 = binary_image(imagex,150)
image3 = binary_image(imagex,200)
```

```python
plt.figure(figsize=(12,8))
images = [image,image1,image2,image3]
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i],cmap='gray')
plt.show()
```
