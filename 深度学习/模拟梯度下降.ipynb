```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-1,1.1,0.1)
y = x**2
```

```python
plt.figure(figsize=(12,6))
plt.plot(x,y,c='b')
plt.show()
```

```python
def derivatives(x):
    return 2*x
```

```python
iters = 200

w = 1
lr = 0.1
ws = [1.0]

for i in range(iters):
    
    grad = derivatives(w)
    w = w - lr * grad
    ws.append(w)
    if abs(grad) < 0.0001:
        print(f"第{i}次迭代达到最低点")
        break
```

```python
temp = [x**0.5 for x in ws]
```

```python
plt.figure(figsize=(12,6))
plt.plot(x,y,c='b')
plt.plot(temp[:i],ws[:i],c='r')
plt.scatter(temp[:i],ws[:i],c='g')
plt.scatter(temp[i],ws[i],c='r')
plt.show()
```

```python
def derivatives2(w1,w2):
    return 2*w1,2*w2
```

```python
iters = 2000

w1 = 1
w2 = 3.0
lr = 0.4
w1s = [1.0]
w2s = [3.0]
for i in range(iters):
    
    grad1,grad2 = derivatives2(w1,w2)
    
    w1 = w1 - lr * grad1
    w2 = w2 - lr * grad2
    
    w1s.append(w1)
    w2s.append(w2)
    
    if grad1 < 0.000001 or grad2 < 0.000001:
        print(f"第{i}次迭代达到最低点")
        break
```

```python
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

x, y = np.meshgrid(x, y)
z = x**2 + y**2

zs = [x**2 + y**2 for x,y in zip(w1s,w2s)]
fig = plt.figure()

ax = fig.add_subplot(111, projection='3d')
# 使用plot_surface绘制三维曲面图
ax.plot_surface(x, y, z,cmap='viridis',alpha=0.5)
ax.scatter(w1s, w2s, zs, c='b', marker='o')

ax.scatter(w1s[i], w2s[i], zs[i], c='r', marker='x',s=1000)
ax.view_init(elev=50, azim=10)
# 添加标题
ax.set_title('3D Surface Plot')

# 显示图形
plt.show()
```
