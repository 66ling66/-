```python
import numpy as np
```

```python
class Tokenizer_self():
    def __init__(self):
        self.max_size = 3
        self.dictnary = ['研究','命','的','起源','研究生','生命']
        self.ans_forword = []
        self.ans_reverse = []
    def mm(self,text):

        len_text = len(text)   #记录待匹配句子长度
        while len_text > 0:
            divided_text = text[:self.max_size]       #取前i个词，i为字典里最长词的长度
            flag = True
            while divided_text not in self.dictnary:
                if len(divided_text) == 1:        #待匹配字符还剩一个词，且该字典中无该词汇，则匹配失败
                    print(f'{divided_text}----未被匹配')
                    flag = False
                    break
                divided_text = divided_text[:len(divided_text)-1]   #继续匹配 前i-1个词汇
            if flag:
                self.ans_forword.append(divided_text)      #将匹配成功的词汇保存起来  这里未匹配成功的词汇将不会保存在这
            text = text[len(divided_text):]            #奖匹配成功的词汇从句子中去除
            len_text = len(text)
        return self.ans_forword

    def rmm(self,text):

        len_text = len(text)
        while len_text > 0:
            divided_text = text[-self.max_size:]   #获取末尾的词

            while divided_text not in self.dictnary:    #若当前词不在字典里则需要去除首字符继续匹配 若词长为1 仍然不在词典里 则未匹配到
                if len(divided_text) == 1:
                    print(f"{divided_text}----未被匹配")
                    break
                divided_text = divided_text[-(len(divided_text)-1):]   #去除首字符
            self.ans_reverse.append(divided_text)                      #添加匹配成功的字符

            text = text[:-len(divided_text)]
            len_text = len(text)
        # print(f'逆向匹配分词结果：{self.ans_reverse[::-1]}')
        return self.ans_reverse[::-1]
    def double_mm(self,text):
        self.ans_forword = []
        self.ans_reverse = []
        result_for = self.mm(text)
        result_re = self.rmm(text)
        if len(result_for) > len(result_re):
            return result_re
        elif len(result_for) < len(result_re):
            return result_for
        else:
            count_for = 0
            count_re = 0
            for word in result_for:
                if len(word) == 1:
                    count_for += 1
            for word in result_re:
                if len(word) == 1:
                    count_re += 1
            if count_for > count_re:    #返回单字数少的那个
                return  result_re
            else:
                return result_for
```

```python
tokenizer = Tokenizer_self()
sentence = '研究生命的起源我热爱大自然'
result_mm = tokenizer.mm(sentence)
result_mm
```
