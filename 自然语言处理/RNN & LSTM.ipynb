```python
import torch.nn as nn
import torch
```

nn.RNN参数解释，input_size就是输入数据的特征维度，比如某个词的特征向量为 torch.tensor([1,2,3]) 则说明这个词特征维度为3。hidden_size表示隐藏层维度，即将特征维度为input_size映射到hidden_size。num_layer表示这个网络有几层，即经过几次映射，例如，如果num_layer为2，那么就是由$x_1$映射到$h_1$再映射到$h_1^{'}$ 

```python
model = nn.RNN(input_size=3,hidden_size=4,num_layers=1,device='cpu')
```
